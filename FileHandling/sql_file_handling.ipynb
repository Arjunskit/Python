Import

````python
import sqlite3
````

Create/Open Database

````python
conn = sqlite3.connect("company.db")
````

Create Cursor

````python
cursor = conn.cursor()
````

Create Table

````python
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary REAL
)
""")
````

Save Changes

````python
conn.commit()
````

Insert Data

````python
cursor.execute("""
INSERT INTO employees
(name, department, salary)
VALUES (?, ?, ?)
""", ("Arjun", "IT", 50000))
````

Notice the ?.

Never do:

```python
cursor.execute(
    f"INSERT INTO employees VALUES('{name}')"
)

```
This can lead to SQL Injection.

Always use placeholders.

Insert Multiple Rows

````python
employees = [
    ("Rahul", "HR", 45000),
    ("Anu", "Finance", 60000),
    ("John", "IT", 70000)
]

cursor.executemany("""
INSERT INTO employees
(name, department, salary)
VALUES (?, ?, ?)
""", employees)

conn.commit()
````

Read Data

````python
cursor.execute("SELECT * FROM employees")

rows = cursor.fetchall()

for row in rows:
    print(row)
````

Read One Row

````python
cursor.execute("SELECT * FROM employees")

print(cursor.fetchone())
````

Update Data

````python
cursor.execute("""
UPDATE employees
SET salary = ?
WHERE name = ?
""", (55000, "Arjun"))

conn.commit()
````

Delete Data

````python
cursor.execute("""
DELETE FROM employees
WHERE name = ?
""", ("Rahul",))

conn.commit()
````

Filter Data

````python
cursor.execute("""
SELECT *
FROM employees
WHERE salary > ?
""", (50000,))
````

Close Connection

````python
conn.close()
````

| Method              | Purpose                  |
| ------------------- | ------------------------ |
| `sqlite3.connect()` | Open/Create database     |
| `cursor()`          | Create cursor            |
| `execute()`         | Run one SQL statement    |
| `executemany()`     | Run many inserts/updates |
| `fetchone()`        | Read one row             |
| `fetchall()`        | Read all rows            |
| `commit()`          | Save changes             |
| `close()`           | Close database           |
