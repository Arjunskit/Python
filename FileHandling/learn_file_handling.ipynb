## JSON (JavaScript Object Notation)

```python
import json
```

#### **json.dump()**

#### *Writes Python data into a JSON file.*

```python
employee = {"id": 101, "name": "Arjun", "department": "IT"}

with open("employee.json", "w") as file:
    json.dump(employee, file)
```

#### **json.load()**
#### *Reads JSON from a file.*

```python
with open("employee.json", "r") as file:
    employee = json.load(file)


print(employee)
```

#### **json.dumps()**

#### *Converts Python data into a JSON string.*

```python
employee = {
    "name": "Arjun"
}

json_string = json.dumps(employee, indent=4)

print(json_string)
type(json_string)
```

#### **json.loads()**

#### *Converts a JSON string back into Python.*

```python
data = '{"name": "arjun", "department": "IT"}'

employee = json.loads(data)

employee
```

#### **Useful Parameters**

#### *indent* - Makes JSON readable. - json.dump(data, file, indent=4)
#### *sort_keys* - Alphabetically sorts keys. - json.dump(data, file, sort_keys=True)


## **CSV (Comma-Separated Values)**


```python
import csv
```

csv.writer() - Writes rows.

writerow() - Writes one row.



```python


with open("employee.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    
    writer.writerow(["id", "name", "department"])

    writer.writerow([1, "Arjun", "IT"])
```

writerows() - Writes multiple rows

```python


employees = [
    [1, "Arjun", "IT"],
    [2, "Rahul", "HR"],
    [3, "Anu", "Finance"]
]


with open("employee.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["id", "name", "department"])

    writer.writerows(employees)
```

csv.reader()

Reads rows as lists.

```python
with open("employee.csv", "r", newline="") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)

print(reader)

```

DictReader()

```python
with open("employee.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["name"])
```

DictWriter()

Perfect for exporting structured data.

```python
import csv

employees = [
    {"id": 1, "name": "Arjun", "department": "IT"},
    {"id": 2, "name": "Rahul", "department": "HR"}
]

with open("employees.csv", "w", newline="", encoding="utf-8") as file:
    fields = ["id", "name", "department"]

    writer = csv.DictWriter(file, fieldnames=fields)

    writer.writeheader()

    writer.writerows(employees)
```

newline="" -->  This prevents extra blank lines on Windows when writing CSV files.
