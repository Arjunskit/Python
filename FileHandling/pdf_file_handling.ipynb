Install

```python
# pip install reportlab
```

Create PDF

```python
from reportlab.pdfgen import canvas


pdf = canvas.Canvas("helloo.pdf")

pdf.drawString(100, 750, "Hello World!")

```

Change Font

```python
pdf.setFont("Helvetica", 16)
```

Draw Lines

```python
pdf.line(100,680,500,680)
```

Draw Rectangle

```python
pdf.rect(80,600,300,100)
```

Draw Circle

```python
pdf.circle(300, 500, 50)
```

Draw Image

```python
pdf.drawImage("logo.png",50,700,width=80,height=80)
```

Multiple Lines

```python
text = pdf.beginText()

text.setTextOrigin(100,700)

text.textLine("Arjun")

text.textLine("Python Developer")

text.textLine("Kerala")

pdf.drawText(text)
```

Change Text Color

```python
from reportlab.lib.colors import blue

pdf.setFillColor(blue)

pdf.drawString(100,700,"Blue Text")
```

Change Line Color

```python
from reportlab.lib.colors import red

pdf.setStrokeColor(red)

pdf.line(100,650,500,650)
```

Draw Table (Simple)

ReportLab's higher-level API (platypus) is best for tables.

```python
from reportlab.platypus import Table

data = [
    ["Name","Salary"],
    ["Arjun",50000],
    ["Rahul",45000]
]

table = Table(data)
```

Open a PDF

```python
import fitz

doc = fitz.open("resume.pdf")

print(doc)
```

Number of Pages

```python
len(doc)

#or

doc.page_count
```

Access a Page

```python
page1 = doc[0]
page2 = doc[1]


last_page = doc[-1]
```

Extract Text

```python
page = doc[0]

text = page.get_text()

print(text)
```

Read Entire PDF

```python
for page in doc:
    print(page.get_text())
```

Search Text

```python
page = doc[0]

results = page.search_for("Python")

print(results)


'''
Output example


[
 Rect(72,120,150,140),
 Rect(72,300,150,320)
]


Each Rect gives the coordinates where the word appears.

'''
```

Highlight Text

```python
page = doc[0]

results = page.search_for("Python")

for rect in results:
    page.add_highlight_annot(rect)

doc.save("highlighted.pdf")
```

Add Text

```python
page.insert_text(
    (100,100),
    "Generated using Python"
)

doc.save("new.pdf")
```

Insert Image

```python
page.insert_image(
    fitz.Rect(50,50,150,150),
    filename="logo.png"
)
```

Extract Images

```python
page = doc[0]

images = page.get_images()

print(images)
```

Save images:

```python
for img in images:

    xref = img[0]

    pix = fitz.Pixmap(doc, xref)

    pix.save(f"{xref}.png")
```

Render PDF as Image

```python
page = doc[0]

pix = page.get_pixmap()

pix.save("page1.png")
```

Rotate Page

```python
page.set_rotation(90)

doc.save("rotated.pdf")
```

Delete Page

```python
doc.delete_page(2)

doc.save("updated.pdf")
```

Insert Blank Page

```python
doc.new_page()

doc.save("newpage.pdf")
```

Merge PDFs

```python
doc1 = fitz.open("resume.pdf")

doc2 = fitz.open("certificate.pdf")

doc1.insert_pdf(doc2)

doc1.save("portfolio.pdf")
```

Metadata

```python
print(doc.metadata)


'''
{
 'author':'Arjun',
 'title':'Resume'
}
'''
```

Close Document

```python
doc.close()
```

Read Every Page

When Should You Use Each Library?

#### **pypdf**
Use when you need to:


- Merge PDFs
- Split PDFs
- Encrypt PDFs
- Rotate pages
- Rearrange pages

**reportlab**

Use when you need to generate:

- Invoices
- Salary slips
- Certificates
- Reports
- Offer letters

**PyMuPDF**

Use when you need to:

- Extract text quickly
- Highlight words
- Search documents
- Insert text/images
- Render PDF pages as images
- Edit existing PDFs

**pdfplumber**

Use when you need to:

- Extract tables
- Read invoices
- Read bank statements
- Parse structured documents
