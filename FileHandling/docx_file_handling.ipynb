Create a Word Document

```python
from docx import Document

document = Document()

document.save("employee.docx")
```

Add a Heading

```python
document.add_heading("Employee Report", level=1)
document.save("employee.docx")
```

Add a Paragraph

```python
document.add_paragraph(
    "This document contains employee information."
)

document.save("employee.docx")
```

Add Multiple Paragraphs

```python
document.add_paragraph("Name : Arjun")

document.add_paragraph("Department : IT")

document.add_paragraph("Salary : 50000")
```

Styling Text with Run

```python
paragraph = document.add_paragraph()

paragraph.add_run("Name : ")

bold_text = paragraph.add_run("Arjun")
bold_text.bold = True
```

Italic

```python
run = paragraph.add_run("Python Developer")

run.italic = True
```

Underline

```python
run = paragraph.add_run("Employee ID")

run.underline = True
```

Add a Table

```python
table = document.add_table(rows=1, cols=3)

header = table.rows[0].cells

header[0].text = "ID"
header[1].text = "Name"
header[2].text = "Department"
```

Add Data to Table

```python
row = table.add_row().cells

row[0].text = "1"
row[1].text = "Arjun"
row[2].text = "IT"

row = table.add_row().cells

row[0].text = "2"
row[1].text = "Rahul"
row[2].text = "HR"
```

Insert Image

```python
from docx.shared import Inches

document.add_picture(
    "logo.png",
    width=Inches(2)
)
```

Page Break

```python
document.add_page_break()
```

Reading an Existing Word File

```python
from docx import Document

document = Document("employee.docx")

for paragraph in document.paragraphs:
    print(paragraph.text)
```

Reading Tables

```python
for table in document.tables:
    for row in table.rows:
        print([cell.text for cell in row.cells])
```
