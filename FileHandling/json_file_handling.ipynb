## JSON (JavaScript Object Notation)

```python
import json
```

#### **json.dump()**

#### *Writes Python data into a JSON file.*

```python
employee = {"id": 101, "name": "Arjun", "department": "IT"}

with open("employee.json", "w") as file:
    json.dump(employee, file)
```

#### **json.load()**
#### *Reads JSON from a file.*

```python
with open("employee.json", "r") as file:
    employee = json.load(file)


print(employee)
```

#### **json.dumps()**

#### *Converts Python data into a JSON string.*

```python
employee = {
    "name": "Arjun"
}

json_string = json.dumps(employee, indent=4)

print(json_string)
type(json_string)
```

#### **json.loads()**

#### *Converts a JSON string back into Python.*

```python
data = '{"name": "arjun", "department": "IT"}'

employee = json.loads(data)

employee
```

#### **Useful Parameters**

#### *indent* - Makes JSON readable. - json.dump(data, file, indent=4)
#### *sort_keys* - Alphabetically sorts keys. - json.dump(data, file, sort_keys=True)
