Create an Excel File  - Workbook()

```python
from openpyxl import Workbook


# create workbook
workbook = Workbook()


# Get active sheet
sheet = workbook.active

# Rename sheet name
sheet.title = "Employees"

# Save workbook
workbook.save("employees.xlsx")

print("Workbook created successfully")
```

Adding Data  - sheet.append()

```python
from openpyxl import Workbook

workbook = Workbook()
sheet = workbook.active
sheet.title = "Employees"

# Header
sheet.append(["ID", "Name", "Department", "Salary"])

# Data
sheet.append([1, "Arjun", "IT", 50000])
sheet.append([2, "Rahul", "HR", 45000])
sheet.append([3, "Anu", "Finance", 60000])


workbook.save("employees.xlsx")
```

Reading an Excel File - load_workbook()

```python
from openpyxl import load_workbook

workbook = load_workbook("employees.xlsx")

sheet = workbook["Employees"]

for row in sheet.iter_rows(values_only=True):
    print(row)
```

Reading a Single Cell

```python
sheet["B2"].value
```

```python
sheet.cell(row=2, column=2).value
```

Updating a Cell

```python
sheet["C2"] = "Software Engineer"

workbook.save("employees.xlsx")
```

Creating Multiple Sheets

```python
marks_sheet = workbook.create_sheet("Marks")

marks_sheet.append(["Name", "Python", "SQL"])

marks_sheet.append(["Arjun", 95, 90])

marks_sheet.append(["Rahul", 88, 85])

workbook.save("employees.xlsx")
```

Getting Sheet Names

```python
workbook.sheetnames
```

Finding Total Rows and Columns

```python
sheet.max_row
```

```python
marks_sheet.max_column
```

Iterating Rows

```python
for row in marks_sheet.iter_rows(values_only=True):
    print(row)
```

Iterating Columns

Iterating Columns

```python
for col in marks_sheet.iter_cols(values_only=True):
    print(col)
```

Deleting a Sheet

```python
del workbook["Marks"]

workbook.save("employees.xlsx")
```

Renaming a Sheet

```python
sheet.title = "Staff"

workbook.save("employees.xlsx")
```
