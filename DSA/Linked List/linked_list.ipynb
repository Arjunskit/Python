```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# '''
# 10 → 20 → 30 → None

# '''

n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.next = n2
n2.next = n3

cursor = n1

while cursor:
    print(cursor.data)
    cursor = cursor.next
```

Doubly linked List

```python
class Node:
    def __init__(self, data):
        self.prev = None
        self.data = data
        self.next = None

n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.next = n2
n2.prev = n1

n2.next = n3
n3.prev = n2
```
