## What is an Array?

An array is a collection of elements stored **contiguously** in memory.

Example:

````python
arr = [10, 20, 30, 40, 50]
````

Visual representation:

```
Index:   0    1    2    3    4
        +----+----+----+----+----+
Array:  |10  |20  |30  |40  |50  |
        +----+----+----+----+----+
```

Each element has an **index**, starting from `0`.

---

## Why Arrays Are Fast

Suppose the first element starts at memory address `1000`, and each integer occupies 4 bytes.

```
Address   Value
1000      10
1004      20
1008      30
1012      40
1016      50
```

To access `arr[3]`, Python (and most languages internally) can calculate:

```
address = base + index × size
```

So it jumps directly to the correct location.

That's why:

```python
arr[3]
```

takes **O(1)** time.

---

## Common Operations

| Operation                               | Time Complexity |
| --------------------------------------- | --------------- |
| Access by index                         | O(1)            |
| Update by index                         | O(1)            |
| Traverse                                | O(n)            |
| Search (unsorted)                       | O(n)            |
| Insert at end (average for Python list) | O(1)            |
| Insert at beginning                     | O(n)            |
| Delete from beginning                   | O(n)            |
| Delete from end (average)               | O(1)            |

# Arrays: Traversal

Traversal means **visiting each element exactly once**.

### Method 1: For-each Loop (Most Pythonic)

```python
arr = [10, 20, 30, 40]

for num in arr:
    print(num)
```

Output:

```
10
20
30
40
```

**Time:** O(n)
**Space:** O(1)

---

### Method 2: Using Indices

```python
arr = [10, 20, 30, 40]

for i in range(len(arr)):
    print(arr[i])
```

This is useful when you need the index.

Output:

```
10
20
30
40
```

---

### Method 3: Index and Value Together

```python
arr = [10, 20, 30, 40]

for index, value in enumerate(arr):
    print(index, value)
```

Output:

```
0 10
1 20
2 30
3 40
```

This is the preferred approach when you need both.

# Common Interview Pattern 1: Find the Maximum Element

### Problem

Find the largest element in an array.

Example:

```python
arr = [12, 45, 8, 90, 23]
```

Expected output:

```
90
```

---

### Brute Force Idea

Compare every element with every other element.

Time: **O(n²)** ❌

Not efficient.

---

### Optimal Idea

Keep track of the largest number seen so far.

```python
def find_max(arr):
    maximum = arr[0]

    for num in arr:
        if num > maximum:
            maximum = num

    return maximum
```

### Dry Run

Array:

```python
[12, 45, 8, 90, 23]
```

| Current | Maximum |
| ------- | ------- |
| 12      | 12      |
| 45      | 45      |
| 8       | 45      |
| 90      | 90      |
| 23      | 90      |

Answer: **90**

### Complexity

* **Time:** O(n)
* **Space:** O(1)
